```python
from nltk.corpus import stopwords

# 1. See which languages have stopword lists
print(stopwords.fileids()[:10])

# 2. Load English stopwords
english_stopwords = stopwords.words('english')
print(f"Count: {len(english_stopwords)}")
print(english_stopwords[:20])

# 3. Check if a specific word is a stopword
print('the' in english_stopwords)
print('python' in english_stopwords)

# 4. Real test: remove stopwords from a sentence
sentence = "This is a simple example showing how stopwords can be removed from a sentence"
words = sentence.lower().split()
filtered = [w for w in words if w not in english_stopwords]

print("Original:", words)
print("Filtered:", filtered)
```
