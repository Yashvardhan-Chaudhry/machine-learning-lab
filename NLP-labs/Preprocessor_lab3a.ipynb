Importing and storing data in Corpus using PlaintextCorpusReader (wrapper)


```python
import nltk
path=r"C:\Users\chaud\Downloads\imdb_text_dataset-20260817T104927Z-1-001\imdb_text_dataset"
nltk.data.path.append(path)  # NLTK 3.10+ security allowlist requires the root to be a known data path
from nltk.corpus import PlaintextCorpusReader
imdb_corpus=PlaintextCorpusReader(path,'.*') #".*" allows all type of files to be read. 
```

Storing corpus into Dataframe.

```python
#print(imdb_corpus.raw('review_1.txt')) # prints raw txt file. 
import pandas as pd
document=[]
for filenames in imdb_corpus.fileids():
    document.append(imdb_corpus.raw(filenames))
df=pd.DataFrame(document, imdb_corpus.fileids())
print(df)
```

Manual OS approach. Same result but more control

```python
import os 
filenames1=os.listdir(path)
documents1=[]
path=r"C:\Users\chaud\Downloads\imdb_text_dataset-20260817T104927Z-1-001\imdb_text_dataset"
for filename in filenames1:
    f=open(path+"//"+filename,'r')# whats double slash for, here making a file object. a pointer to the file. mode is r
    documents1.append(f.read())
    f.close()
#documents1[0][:10]
```

NORMALIZATION + TOKENIZATION

```python
# PREPROCESSING. NORMALIZING AND THEN TOKENIZING
from nltk import word_tokenize # uses pkunt_tab ML model for tokenization.
normalize_tokenize=[]
for doc in documents1:
    normalize_tokenize.append([word.lower() for word in nltk.word_tokenize(doc) if word.isalpha()]) 
    # above statement, tokenization done through wrapper word_tokenize. Normalization done here is 1. changed case, 2.Only alphabets.
```

```python
normalize_tokenize[0]
```

Now tokenized. removing stopwords


```python
from nltk.corpus import stopwords
stopword_list=stopwords.words('english')
nostopword=[]
for doc in normalize_tokenize:
    nostopword.append([word for word in doc if word not in stopword_list])
```

```python
# dimensionality reduction displayed 
print(nostopword[0])
print(f"Length before:{len(normalize_tokenize)}")
print(f"Length after: {len(nostopword)}")
print("These were lengths of document thats why they are same haha")
print(f"Length before:{len(normalize_tokenize[0])} and Length after: {len(nostopword[0])}")
```

Morphological Analysis: Stemmers


```python
from nltk.stem import PorterStemmer
ps=PorterStemmer()
preprocessed_text=[]
#dir(ps)
for doc in nostopword:
    preprocessed_text.append(' '.join([ps.stem(word) for word in doc]))
    
```

Final comparison between preprocessed and raw text.

```python
print(documents1[0])
print(len(documents1[0]))
print(preprocessed_text[0])
len(preprocessed_text[0])
```

Testing different Stemmers.

```python
from nltk.stem import LancasterStemmer
ls=LancasterStemmer()
preproc_ls=[]
for doc in nostopword:
    preproc_ls.append(' '.join([ls.stem(word) for word in doc]))
print(f"Length of Lancaster:{len(preproc_ls[0])} and Length of Porter:{len(preprocessed_text[0])} ")
print(preprocessed_text[0])
print(preproc_ls[0])
```
