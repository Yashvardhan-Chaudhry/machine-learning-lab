# Converting Corpus to Features



In order to convert corpus to feature matrix, following steps are used:
1. Loading your own corpus
2. Pre-processing corpus: Normalization, Tokenization, Stop-word Removal, Stemming
3. Converting pre-processed corpus to feature matrix (TDM/DTM or TTM)

# Loading your own corpus
User defined corpus can be imported in Python using two methods:
1. Using nltk.corpus PlainTextCorpusReader or CategorizedCorpusReader
2. Using file method of Python

# CorpusReader





```python
import nltk
from nltk.corpus import PlaintextCorpusReader
path=r"C:\Users\jasme\Downloads\imdb_text_dataset"
dataset=PlaintextCorpusReader(path,'.*')
# The same dataset can be downloaded fom following path: https://drive.google.com/drive/folders/1f5y9XLsoV9mt_j0O1k-xv0UvE14cu6-7
```

```python
#Converting the dataset to a list (where each string represents a seperate document)
corpus=[]
for filename  in dataset.fileids():
    corpus.append(dataset.raw(fileids=filename))
corpus[0]
```

# Loading Corpus Using File Method of Python

Using files:  
File_object=open(r"File_Name","Access_Mode")

Access Modes :
1. Read Only (‘r’)
2. Read and Write (‘r+’)
3. Write Only (‘w’)
4. Write and Read (‘w+’)
5. Append Only (‘a’)
6. Append and Read (‘a+’)

```python
import os
filenames=os.listdir(path)
corpus=[]
for filename  in filenames:
    f=open(path+'//'+filename,'r')
    corpus.append(f.read())
    f.close()
corpus[0]
```

# Pre-processing Step 1 and 2: Normalization and Tokenization

Normalization involves removing special symbols, URLs, HTML tags, and converting text to lower case
Tokenization involves converting each document as list of words. It can be done in two ways:
1. .split() method of list
2. word_tokenize method of nltk.tokenize

```python
#Tokenization using .split()
normalize_tokenize=[]
for doc in corpus:
    normalize_tokenize.append([word.lower() for word in doc.split() if word.isalpha()])
normalize_tokenize[0]
```

```python
#Tokenization using word_tokenize

normalize_tokenize1=[]
for doc in corpus:
    normalize_tokenize1.append([word.lower() for word in nltk.word_tokenize(doc) if word.isalpha()])
normalize_tokenize1[0]
```

# Pre-processing Step 3: Stop-word Removal
A stop word is a commonly used word (such as “the”, “a”, “an”, “in”) that does not have any linguistic importance in NLP applications

NLTK(Natural Language Toolkit) in python has a list of stopwords stored in stopwords corpus in 16 different languages.

The name of fields is the name of language.

```python
import nltk
from nltk.corpus import *
stopword_list=nltk.corpus.stopwords.words('english') #stopword will contain list of all stopwords of english language
no_stop=[]
for doc in normalize_tokenize:
    no_stop.append([word for word in doc if word not in stopword_list])
no_stop[0]
```

# Pre-processing Step 4: Stemming

Stemming is a process that maps variant word forms to their base forms (play, plays, playing, played )

nltk.stem has number of stemming algorithms named as "PorterStemmer", "LancasterStemmer", etc. These algorithms accepts the list of tokenized word and stems it into root word.


```python
#Stemming Example
from nltk.stem import PorterStemmer #Importing porter stemmer class
ps=PorterStemmer() #Creating an object of PorterStemmer Class
ps.stem('departed') #stemming a word using .stem method
```

```python
#Stemming the corpus
preprcossed_text=[] #will contain final pre-processed documents
from nltk.stem import PorterStemmer
ps=PorterStemmer()
for doc in no_stop:
    preprcossed_text.append(' '.join([ps.stem(word) for word in doc]))
preprcossed_text[0]
```

```python
#Lemmatizing the corpus
preprcossed_text1=[] #will contain final pre-processed documents
from nltk.stem import WordNetLemmatizer
wn=WordNetLemmatizer()
for doc in no_stop:
    preprcossed_text1.append(' '.join([wn.lemmatize(word) for word in doc]))
preprcossed_text1[0]
```

```python
#Storing corpus and preprocessed corpus in DataFrame
import pandas as pd
df=pd.DataFrame({'reviews':corpus,'preprocessed_reviews':preprcossed_text})
```
