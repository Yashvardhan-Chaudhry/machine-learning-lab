Web Scraping: Beautiful Soup

Fetches html content, beautiful soup makes a parse tree consisting of tags, we iterate through these tags fo fetch data


```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
def get_data(pageno):
    url='https://www.amazon.in/gp/bestsellers/books/ref=zg_bs_pg_'+str(pageno)+'_books?ie=UTF8&pg='+str(pageno)
    html=requests.get(url)
    soup=BeautifulSoup(html.content,'html.parser')
    data=[]
    for cell in soup.find_all('div',{'class':"zg-grid-general-faceout"}):
        book_name=cell.find('div',{'class':"_cDEzb_p13n-sc-css-line-clamp-1_1Fn1y"})
        #print(book_name.text if book_name else 'NaN')#checked it works yay 
        #author_name=cell.find_all[1]('div',{'class':"_cDEzb_p13n-sc-css-line-clamp-1_1Fn1y"}) #this doesnt work with find, because the class is exactly the same as bookname. Now ill try to find some parent tag. OR use find_all 
        authorname=cell.find('div',{'class':"a-row a-size-small"})
        stars=cell.find('span',{'class':"a-icon-alt"})
        votes=cell.find('span',{'class':"a-size-small"})
        price=cell.find('span',{'class':"p13n-sc-price"}) #_cDEzb_p13n-sc-price_3mJ9Z price class was this removed dynamic part 
        #print(author_name.text if author_name else "NaN")
        #print(authorname.text if authorname else "NaN")
        #print(stars.text if stars else "NaN")
        #print(votes.text if votes else "NaN")
        #print(price.text if price else "NaN")
        book_link=cell.find('a',{'class':"a-link-normal aok-block"})
        #print('www.amazon.in'+book_link['href'] if book_link else 'Nan') # added www to add base url. href as anchor (a) doesnt have text. 
        data.append({'Book Name':book_name.text if book_name else 'NaN',
                     'Author Name':authorname.text if authorname else 'NaN',
                     'Ratings': stars.text if stars else 'NaN',
                     'Votes': votes.text if votes else 'NaN',
                     "Book_Link": 'www.amazon.in'+book_link['href'] if book_link else 'NaN'})
        #print(f"Star rating:{stars.text}\n Votes: {votes.text}\n Price:{price.text}") doesnt work 
    return data
```

```python
data=get_data(1)
df=pd.DataFrame(data)
df
#soup.find('title').text #.text/.string= gives you text inside the html tag. Find= returns first occurence of a tag you want.   .
#soup.find_all('div') #find all gives all occurence. Find gives first occurence
```

```python
no_of_pages=2
all_data=[]
for pageno in range (1,no_of_pages+1):
    all_data.extend(get_data(pageno))
```
