```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
```

```python
dataset=load_iris()
df=pd.DataFrame(dataset.data, columns=dataset.feature_names)
df['target']=dataset.target # adding label column at the end so iloc[:,:-1] gives features
df.head()
```

```python
X=df.iloc[:,:-1]
Y=df.iloc[:,-1]
```

```python
def between_class(X,Y):
    global_mean=X.mean(axis=0)
    classes=np.unique(Y)
    n,k=X.shape
    Sb=np.zeros((k,k),dtype=float)
    for cls in classes:
        mask=(Y==cls)
        X_subset=X[mask]
        local_mean=X_subset.mean(axis=0)
        diff=local_mean-global_mean
        diff=diff.values.reshape(-1,1) # column vector (k,1) so that diff@diff.T is (k,k)
        Sb+=X_subset.shape[0]*(diff@diff.T) # N_c * outer product
    return Sb
#try with numpy.outer too OR make into row matrices and do transpose diff into diff.
```

```python
Sb=between_class(X,Y)
```

```python
def within_class(X,Y):
    n,k=X.shape
    Sw=np.zeros((k,k),dtype=float)
    classes=np.unique(Y)
    for cls in classes:
        mask=(Y==cls)
        X_subset=X[mask]
        local_mean=X_subset.mean(axis=0)
        diff=X_subset-local_mean
        Sw+=diff.T@diff
    return Sw
```

```python
Sw=within_class(X,Y)
```
