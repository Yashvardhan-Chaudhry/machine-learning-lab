PCA from scratch on Iris

```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
```

```python
dataset=load_iris()
df=pd.DataFrame(dataset.data, columns=dataset.feature_names)
df['target']=dataset.target # adding label column at the end so iloc[:,:-1] gives features
df.head()
```

```python
X=df.iloc[:,:-1]
Y=df.iloc[:,-1]
```

Standardize. PCA is sensitive to scale, so mean=0 std=1 for each feature

```python
from sklearn.preprocessing import StandardScaler
X1=StandardScaler().fit_transform(X)
```

Covariance matrix. Since data is already centred, cov = (X^T X)/(n-1)

```python
cov=(X1.T@X1)/(len(X1)-1) # sir wrote /len(X1)-1 which subtracts 1 from every entry, brackets needed
cov
```

Eigen decomposition. eigh is for symmetric matrices, returns eigenvalues in ascending order

```python
values, vectors=np.linalg.eigh(cov)
```

```python
values
```

Explained variance = eigenvalue / sum of eigenvalues (in %)

```python
explained_variance=[]
for value in values:
  explained_variance.append((value*100)/sum(values))
```

```python
explained_variance
```

Projection. Last 2 columns of vectors are the top 2 principal components (ascending order)

```python
X_new=X1@vectors[:,2:4]
X_new[:5]
```

```python
df_reduced=pd.DataFrame(X_new,columns=['X1','X2'])
df_reduced['target']=df['target']
df_reduced.head()
```

Correlation between the new features should be ~0 (components are orthogonal)

```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.heatmap(df_reduced.iloc[:,:-1].corr(),annot=True)
plt.show()
```

```python
sns.scatterplot(x='X1',y='X2',hue='target',data=df_reduced)
plt.show()
```

In-built method: sklearn PCA

```python
from sklearn.decomposition import PCA
pca=PCA(n_components=2)
X_pca=pca.fit_transform(X1)
X_pca[:5]
```

```python
pca.explained_variance_ratio_*100 # same as last 2 values of explained_variance above, just descending
```

```python
pca.components_ # rows are the principal components, same as vectors[:,2:4].T (sign may flip)
```

```python
df_pca=pd.DataFrame(X_pca,columns=['PC1','PC2'])
df_pca['target']=df['target']
sns.scatterplot(x='PC1',y='PC2',hue='target',data=df_pca)
plt.show()
```
