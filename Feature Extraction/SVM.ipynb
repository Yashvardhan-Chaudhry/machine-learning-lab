```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("data/user_data.csv") # relative to Feature Extraction/, run the notebook from there
```

```python
df.shape
```

```python
R=df.pivot_table(index='user_id', columns='movie_id',values='rating',fill_value=0)
```

```python
R.shape
```

```python
values, vectors=np.linalg.eigh(R.T@R) #claude when to use eigh and when to use eig?
index=np.argsort(values)[::-1]
vectors=vectors[:,index]
V=vectors[:,:50]
values=values[index]
S=np.maximum(np.sqrt(values),0)[:50]
U=(R@V)/S
```

```python
R_low=U@np.diag(S)@V.T
R_low
```

```python
def recommendation(user_id,topn):
    user_index=R.index.get_loc(user_id)
    scores=R_low.iloc[user_index]
    scores[(R.iloc[user_index]>0).to_numpy()]=np.inf
    sorted_indices=np.argsort(scores)[::-1]
    top_indices= sorted_indices[:topn]
    recommendation=pd.DataFrame({'movie_id':R.columns[top_indices],'scores':scores[top_indices]})

    return recommendation
```

```python
recommendation(1,10)
```
