```python
import numpy as np
import cv2
```

```python
img=cv2.imread(r"C:\Users\chaud\Downloads\rgb-grayscale.jpg")
b,g,r= cv2.split(img)
wb, wg,wr= 0.3,0.3,0.4
if abs(wb+wg+wr-1.0)>1e-6:
    print ("Errors: weights sum not one.")
```

```python
gray=(r*wr+b*wb+g*wg).astype(np.uint8)
```

```python
cv2.imshow("Orginal",img)
cv2.imshow("60s",gray)
cv2.waitKey(0) 
cv2.destroyAllWindows()
```
