Grayscale --> Binary image 

```python
import numpy as np
import cv2
img=cv2.imread(r"C:\Users\chaud\Downloads\rgb-grayscale.jpg",cv2.IMREAD_GRAYSCALE)
if img is None: 
    print("Error: 404")

```

Taking mean intensity as threshold 


```python
thresh=np.mean(img)
#mask=img>=thresh
#using where
mask=np.where(img>=thresh,255,0) 
cv2.imshow("Grayscale",img)
cv2.imshow("Binary",mask.astype(np.uint8)) # Learning: First converted grayscale into true or false. Then converted those true and False into 1 and 0 using astype. Now coverting those 1 into 255 as its a 8 bit picture and 255 technically means 1.
cv2.waitKey(0) 
cv2.destroyAllWindows()
```

Taking input threshold


```python
thresh = int(input("Enter the threshold intensity: "))
mask=img>=thresh
cv2.imshow("Grayscale",img)
cv2.imshow("Binary", mask.astype(np.uint8)*255)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
