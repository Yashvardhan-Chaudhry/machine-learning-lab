```python
import cv2
import numpy as np
img=cv2.imread(r"C:\Users\chaud\Downloads\rgb-grayscale.jpg",cv2.IMREAD_GRAYSCALE)
if img is None: 
    print("Error: 404")
```

Question1. Enhance image using Log transform


```python
c=1
s= c*np.log(1+img.astype(np.float32))# log (1+x) #transform

log_img= ((s - s.min())/(s.max()- s.min()))*255 #min max norma- scaling 
log_img=log_img.astype(np.uint8) #type conversion
```

Question 2: Power law/ Gamma Transform

```python
c=1
gamma=float(input("Enter gamma: "))
temp_img=img.astype(np.float32) #type cast 
temp_img/=255 #normalize
s=c*(temp_img**gamma) #transform
gamma_img= (s*255).astype(np.uint8) #scaled back and type casted 

```

Question3: Contrast Stretching

```python
if img is None: 
    print("Error: 404")
else:
    r_min=int(input("Enter input lower intensity: "))
    r_max=int(input("Enter input higher intensity: "))
    s_min=int(input("Enter output min: "))
    s_max=int(input("Enter output max: "))
if r_min>=r_max:
    print("Error in input range")
    exit()
elif s_min>=s_max:
    print("Error in output range")
    exit()
else:
    img_float=img.astype(np.float32)
mask=(img_float>=r_min) & (img_float<=r_max )#or simply just use s=np.clip(img_float,r_min,r_max)
stretched= ((img_float - r_min)/(r_max - r_min))*(s_max - s_min)+s_min #transform 
s=np.where(mask,stretched,img_float) # where(cond,true,false)
s=s.astype(np.uint8)
print(s.dtype, s.min(), s.max())
```

Display

```python
cv2.imshow("Log Transformed",log_img)
cv2.imshow("Gamma Transformed",gamma_img)
cv2.imshow("Constrast Stretching",s)
cv2.imshow("Orginal",img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
