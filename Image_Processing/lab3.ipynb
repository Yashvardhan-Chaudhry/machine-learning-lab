Padding an image with border.

```python
import cv2
import numpy as np
img=cv2.imread(r"C:\Users\chaud\Downloads\rgb-grayscale.jpg",cv2.IMREAD_GRAYSCALE)
```

INPUT

```python
if img is None: 
    print("Error: 404")
else:
    width=int(input("Enter width of the padding: "))
    value=int(input("Enter the intensity of padding: "))
```

Grayscale + Complement with padding. 

```python
padded=cv2.copyMakeBorder(img, width,width, width, width, cv2.BORDER_CONSTANT,value=value)
complement= 255-img
complement_padded= cv2.copyMakeBorder(complement,width,width,width,width, cv2.BORDER_CONSTANT,value=value)
#Display
cv2.imshow("Complement",complement)
cv2.imshow("Complement_Padded",complement_padded)
cv2.imshow("Orginal",img)
cv2.imshow("Padded",padded)
cv2.waitKey(0) 
cv2.destroyAllWindows()
```

Binary image and its complement with padding


```python
mask= (img>=np.mean(img)).astype(np.uint8)*255
complement_mask=255-mask
mask_padded=cv2.copyMakeBorder(mask,width,width,width,width,cv2.BORDER_CONSTANT,value=value)
#display
cv2.imshow("Binary",mask)
cv2.imshow("Complement",complement_mask)
cv2.imshow("Padded Binary",mask_padded)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
