```python
import numpy as np
import cv2
path=r"C:\Users\chaud\Downloads\salt_pepper_duck.png"
img= cv2.imread(path,cv2.IMREAD_GRAYSCALE)
```

INPUT 

```python
if img is None:
    print("Error 404")
else:
    n=int(input("\n1.Replicate Padding. \n2. Zero Padding. \n Enter padding method:"))
    filter_size=int(input("Enter the filter size: "))
    if filter_size<3 or filter_size%2==0:
        print("Filter size is invalid")
        exit()
```

Replicate border padding


```python
if n==1:
    border_type= cv2.BORDER_REPLICATE
elif n==2:
    border_type=cv2.BORDER_CONSTANT
else: 
    print("Invalid Value entered")
```

Averaging Filter

```python
averaging_kernel= np.ones((filter_size,filter_size),dtype=np.float32)
#Normalize
averaging_kernel/=averaging_kernel.sum()
print(f"Averaging Kernel:\n{averaging_kernel}")
# Apply slide slide
average_result=cv2.filter2D(img,-1,averaging_kernel,borderType= border_type )#claude what to type here?
```

Weighted Filter

```python
#Generate the kernel.
kernel= np.zeros((filter_size,filter_size),dtype=np.float32)
centre=filter_size//2
for i in range (filter_size):
    for j in range(filter_size):
        x= centre - i 
        y= centre - j
        kernel[i,j]=max(abs(x),abs(y))
#normalize
kernel/=kernel.sum()
print(f"Weighted Kernel:\n{kernel}")
#apply
Weight_result=cv2.filter2D(img,-1,kernel,borderType=border_type)
```

Gaussian Filter

```python
#Generate kernel
kernel= np.zeros((filter_size,filter_size),dtype=np.float32)
sigma=float(input("Enter Sigma: "))
centre= filter_size//2
for i in range (filter_size):
    for j in range(filter_size):
        x= centre-i 
        y=centre - j
        kernel[i,j]=1/(2*np.pi*sigma**2 )* np.exp(-(x**2+y**2)/(2*sigma**2))
#normalize
kernel/=kernel.sum()
print(f"Gaussian Kernal: \n {kernel}")
#apply
Gauss_result=cv2.filter2D(img,-1,kernel,borderType=border_type)
```

Display

```python

cv2.imshow("Averaging Filter",average_result)
cv2.imshow("Orginal",img)
cv2.imshow("Weighted Filter",Weight_result)
cv2.imshow("Gaussian Filter", Gauss_result)
cv2.waitKey(0) 
cv2.destroyAllWindows()
```
