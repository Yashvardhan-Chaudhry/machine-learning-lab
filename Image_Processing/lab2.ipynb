RGB--> Grayscale

```python
import numpy as np
import cv2
img=cv2.imread(r"C:\Users\chaud\Downloads\rgb-grayscale.jpg")
if img is None:
    print("Error:404")
b,g,r=cv2.split(img)
```

Case 1: By taking input weights

```python
wb=float(input("Enter weights of blue: "))
wg=float(input("Enter weights of Green: "))
wr=float(input("Enter weights Red: "))
if abs(wb+wr+wg-1.0)>1e-4:
    print("Weights dont add upto one")
gray=(wb*b+wg*g+wr*r).astype(np.uint8)
```

Case 2: Taking Mean Average 

```python
gray=((b.astype(np.float16)+g.astype(np.float16)+r.astype(np.float16))/3).astype(np.uint8)
```

Display

```python
cv2.imshow("Original",img)
cv2.imshow("60s",gray)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
